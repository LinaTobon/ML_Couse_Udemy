```python
class Mod():
    def __init__(self,value,modulus):
        if not isinstance(self.value,int):
            raise ValueError('value must be an int')
        if not isinstance(self.modulus,int):
            raise ValueError('modulus must be an int')
        if not self.modulus>0:
            raise ValueError('modulus must be a positive value')
        
        self._modulus = modulus
        self._value = value % modulus

    @property
    def value(self):
        return self._value
    @value.setter
    def value(self,value):
        self._value = value

    @property
    def modulus(self):
        return self._modulus
    
    def __repr__(self):
        print(f'Mod(value:{self._value},modulus:{self._modulus})')
    
    def __int__(self):
        return self.value

    def __eq__(self,other):
        if isinstance(other,int):
            result = other%self.modulus == self.value
        if isinstance(other,Mod):
            if other._modulus == self._modulus:
                result = self.value == other.value
                return result
        else:
            return NotImplemented
    
    def __hash__(self):
        return hash((self.value,self.modulus))
    
    def __neg__(self):
        return Mod(-self.value,self.modulus)

    def __add__(self,other): 
        if isinstance(other,Mod):
            if other.modulus == self.modulus:
                new_value = self.value + other.value
                return Mod(new_value,self.modulus)
        
        if isinstance(other,int):
            new_value = self.value +other
            return Mod(new_value,self.modulus)
        return NotImplemented
    
    def __sub__(self,other):
        if isinstance(other,Mod):
            if other.modulus == self.modulus:
                new_value = self.value - other.value
                return Mod(new_value,self.modulus)
        if isinstance(other,int):
            new_value = self.value - other
            return Mod(new_value,self.modulus)
        return NotImplemented
    
    def __mul__(self,other):
        if isinstance(other,Mod):
            if other.modulus == self.modulus:
                new_value = self.value*other.value
                return Mod(new_value,self.modulus)
        if isinstance(other,int):
            new_value = other*self.value
            return Mod(new_value,self.modulus)
        return NotImplemented
    
    def __pow__(self,other):
        if isinstance(other,Mod):
            if other.modulus == self.modulus:
                new_value = self.value**other.value
                return Mod(new_value,self.modulus)
        if isinstance(other,int):
            new_value = self.value**(other%self.modulus)
            return Mod(new_value,self.modulus)
        return NotImplemented
    
    def __iadd__(self,other): 
        if isinstance(other,Mod):
            if other.modulus == self.modulus:
                self.value = (self.value + other.value)%self.modulus
                return self
        
        if isinstance(other,int):
            self.value = (self.value +other)%self.modulus
            return self
        return NotImplemented
    
    def __isub__(self,other): 
        if isinstance(other,Mod):
            if other.modulus == self.modulus:
                self.value = (self.value - other.value)%self.modulus
                return self
        
        if isinstance(other,int):
            self.value = (self.value - other)%self.modulus
            return self
        return NotImplemented
    
    def __imul__(self,other): 
        if isinstance(other,Mod):
            if other.modulus == self.modulus:
                self.value = (self.value * other.value)%self.modulus
                return self
        
        if isinstance(other,int):
            self.value = (self.value * other)%self.modulus
            return self
        return NotImplemented
    
    def __ipw__(self,other): 
        if isinstance(other,Mod):
            if other.modulus == self.modulus:
                self.value = (self.value * other.value)%self.modulus
                return self
        
        if isinstance(other,int):
            self.value = (self.value ** (other%self.modulus))%self.modulus
            return self
        return NotImplemented
    
    def __lt__(self,other): 
        if isinstance(other,Mod):
            if other.modulus == self.modulus:
                return self.value<other.value
        
        if isinstance(other,int):
            return self.value<(other%self.modulus)
        return NotImplemented
```
